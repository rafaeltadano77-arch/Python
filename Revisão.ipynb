#** Leia as duas notas de um aluno e mostre se ele está aprovado ou reprovado**

```python
n1 = float(input("Digite a nota 1: "))
n2 = float(input("Digite a nota 2: "))

media = (n1 + n2) / 2

print(f"A média do aluno é {media:.1f}")

if media >= 6:
    print("Aprovado")
else:
    print("Reprovado")
```

**Testando se um número é par**

```python
n=int(input("Digite um número: "))
if n % 2 == 0:
  print(f"O número {n} é par")
else:
  print(f"O número {n} é ímpar")
```

```python
senha=0
tent=0
while senha != 1234:
  senha=int(input("Digite a senha: "))
  tent+=1
  if tent==3 and senha != 1234:
    print("Acesso bloqueado")
    break
  if senha == 1234:
    print("Acesso permitido")
```
