```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Frases para treinar o chatbot
Frases = [
    # Iluminação
    "A iluminação pública da minha rua está muito fraca, deixando os moradores inseguros durante a noite.",
    "Os postes de luz do meu bairro estão queimados há vários dias e ninguém veio consertar.",
    "A praça do bairro fica completamente escura depois que anoitece, dificultando a circulação das pessoas.",

    # Esgoto
    "O esgoto está voltando pelos ralos das casas e causando um cheiro muito forte na vizinhança.",
    "Existe um vazamento de esgoto a céu aberto perto da minha residência, trazendo riscos para a saúde dos moradores.",
    "A rede de esgoto do bairro parece estar entupida, causando mau cheiro e desconforto para todos.",

    # Transporte
    "Os ônibus demoram muito para passar e os moradores chegam atrasados ao trabalho e à escola.",
    "Os coletivos estão sempre lotados nos horários de pico e muitos passageiros precisam viajar em pé.",
    "O bairro precisa de mais linhas de ônibus, pois o transporte atual não atende a todos os moradores.",

    # Ruas e calçadas
    "A rua onde moro está cheia de buracos, dificultando a passagem dos veículos e causando acidentes.",
    "As calçadas estão quebradas e dificultam a circulação de idosos, cadeirantes e pessoas com mobilidade reduzida.",
    "O asfalto está muito danificado e precisa de manutenção para melhorar a segurança dos motoristas.",

    # Lixo
    "A coleta de lixo não está acontecendo regularmente e os resíduos ficam acumulados nas calçadas.",
    "Há muito lixo espalhado pelas ruas do bairro, causando mau cheiro e atraindo animais.",
    "O caminhão de coleta não passa há vários dias e os moradores não sabem onde colocar os resíduos.",

    # Alagamento
    "Quando chove forte, a água se acumula nas ruas e invade as casas dos moradores.",
    "Os bueiros estão entupidos e não conseguem escoar a água da chuva, provocando alagamentos.",
    "A falta de drenagem adequada faz com que várias ruas fiquem inundadas durante as tempestades.",

    # Saúde
    "O posto de saúde demora muito para atender os pacientes e está difícil conseguir uma consulta.",
    "Os moradores precisam esperar várias horas para receber atendimento médico no posto de saúde.",
    "Está faltando atendimento médico no bairro e muitas pessoas precisam procurar unidades distantes.",

    # Água
    "O abastecimento de água é interrompido frequentemente e prejudica as atividades diárias das famílias.",
    "Estamos há vários dias com pouca água nas torneiras e precisamos de uma solução urgente.",
    "A água não chega regularmente às casas do bairro, causando dificuldades para cozinhar e fazer a limpeza.",

    # Gestão pública
    "A prefeitura precisa ouvir as reclamações dos moradores e tomar providências para melhorar o bairro.",
    "Já comunicamos vários problemas à administração municipal, mas até agora não recebemos uma solução.",
    "Os espaços públicos estão abandonados e precisamos de mais investimentos na manutenção da cidade."
]

# Categorias correspondentes a cada frase
Categorias = [
    "Iluminação",
    "Iluminação",
    "Iluminação",

    "Esgoto",
    "Esgoto",
    "Esgoto",

    "Transporte",
    "Transporte",
    "Transporte",

    "Ruas",
    "Ruas",
    "Ruas",

    "Lixo",
    "Lixo",
    "Lixo",

    "Alagamento",
    "Alagamento",
    "Alagamento",

    "Saúde",
    "Saúde",
    "Saúde",

    "Água",
    "Água",
    "Água",

    "Gestão pública",
    "Gestão pública",
    "Gestão pública"
]

# Criar o vetorizador
vetorizador = CountVectorizer()
X = vetorizador.fit_transform(Frases)

# Treinar o modelo
modelo = MultinomialNB()
modelo.fit(X, Categorias)

# Iniciar o chatbot
print("PrefeituraBot: Olá! Sou o assistente virtual da prefeitura.")
print("PrefeituraBot: Descreva o problema que você encontrou na cidade.")
print("PrefeituraBot: Digite 'sair' para encerrar.")

while True:
    entrada = input("\nVocê: ").strip()

    if entrada.lower() == "sair":
        print("PrefeituraBot: Obrigado por participar! Até logo.")
        break

    # Transformar a reclamação em números
    entrada_vetorizada = vetorizador.transform([entrada])

    # Verificar se há palavras conhecidas
    if entrada_vetorizada.nnz == 0:
        print("PrefeituraBot: Não consegui identificar o problema. Tente descrever com outras palavras.")
        continue

    # Prever a categoria da reclamação
    previsao = modelo.predict(entrada_vetorizada)[0]

    # Responder de acordo com a categoria
    if previsao == "Iluminação":
        print("PrefeituraBot: Entendi sua reclamação sobre a iluminação pública. O problema pode ser encaminhado ao setor responsável.")

    elif previsao == "Esgoto":
        print("PrefeituraBot: Entendi sua reclamação sobre a rede de esgoto. O problema pode ser encaminhado ao setor de saneamento.")

    elif previsao == "Transporte":
        print("PrefeituraBot: Entendi sua reclamação sobre o transporte público. O problema pode ser encaminhado ao setor de transportes.")

    elif previsao == "Ruas":
        print("PrefeituraBot: Entendi sua reclamação sobre as ruas e calçadas. O problema pode ser encaminhado ao setor de manutenção urbana.")

    elif previsao == "Lixo":
        print("PrefeituraBot: Entendi sua reclamação sobre a coleta de lixo. O problema pode ser encaminhado ao setor de limpeza urbana.")

    elif previsao == "Alagamento":
        print("PrefeituraBot: Entendi sua reclamação sobre alagamentos. O problema pode ser encaminhado ao setor responsável pela drenagem urbana.")

    elif previsao == "Saúde":
        print("PrefeituraBot: Entendi sua reclamação sobre o atendimento de saúde. O problema pode ser encaminhado ao setor de saúde.")

    elif previsao == "Água":
        print("PrefeituraBot: Entendi sua reclamação sobre o abastecimento de água. Entre em contato com o serviço responsável pelo abastecimento.")

    elif previsao == "Gestão pública":
        print("PrefeituraBot: Entendi sua reclamação sobre os serviços públicos. O relato pode ser encaminhado à administração municipal.")

    print("PrefeituraBot: Obrigado por compartilhar sua reclamação!")
```
